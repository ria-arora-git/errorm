```python
import pandas as pd
import seaborn as sns
import numpy as np

df = sns.load_dataset("titanic")

print(df.head())

print("\nMissing Values:\n")
print(df.isnull().sum())

df = df.drop_duplicates()

numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df = df[(df[col] >= lower) & (df[col] <= upper)]

print("\nCleaned Dataset Shape:\n")
print(df.shape)

corr_matrix = df.corr(numeric_only=True)

corr_pairs = corr_matrix.unstack()

corr_pairs = corr_pairs[
    corr_pairs.index.get_level_values(0)
    != corr_pairs.index.get_level_values(1)
]

positive_corr = corr_pairs.sort_values(ascending=False)

negative_corr = corr_pairs.sort_values()

print("\nMost Positively Correlated Attributes:\n")
print(positive_corr.head(1))

print("\nMost Negatively Correlated Attributes:\n")
print(negative_corr.head(1))
```

```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris
import seaborn as sns
import matplotlib.pyplot as plt

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

print("\nMean:\n")
print(df.mean())

print("\nMedian:\n")
print(df.median())

print("\nMode:\n")
print(df.mode().iloc[0])

print("\nStandard Deviation:\n")
print(df.std())

print("\nStandard Error:\n")
print(df.sem())

print("\nConfidence Interval (95%):\n")

for col in df.columns:
    mean = df[col].mean()
    sem = df[col].sem()
    margin = 1.96 * sem

    lower = mean - margin
    upper = mean + margin

    print(col, ":", (lower, upper))

corr_matrix = df.corr()

print("\nCorrelation Matrix:\n")
print(corr_matrix)

plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()

covariance = np.cov(
    df["sepal length (cm)"],
    df["petal length (cm)"]
)

print("\nCovariance Matrix:\n")
print(covariance)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_openml

titanic = fetch_openml(name="titanic", version=1, as_frame=True)

df = titanic.frame

plt.figure(figsize=(8, 6))

sns.countplot(
    data=df,
    x="sex",
    hue="survived"
)

plt.xlabel("Gender")
plt.ylabel("Frequency")
plt.title("Survivors and Non-Survivors by Gender")
plt.legend(title="Survived")
plt.show()

plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x="age",
    y="fare",
    hue="survived"
)

plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Scatter Plot of Age vs Fare")
plt.show()

plt.figure(figsize=(8, 6))

sns.kdeplot(
    data=df,
    x="age",
    fill=True,
    label="Age"
)

sns.kdeplot(
    data=df,
    x="fare",
    fill=True,
    label="Fare"
)

plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Density Distribution of Age and Fare")
plt.legend()
plt.show()

sns.pairplot(
    df[["age", "fare", "pclass", "survived"]],
    hue="survived"
)

plt.show()
```

```python
import pandas as pd
from sklearn.datasets import fetch_openml

titanic = fetch_openml(name="titanic", version=1, as_frame=True)

df = titanic.frame

passengers_less_than_30 = df.groupby(df["age"] < 30).size()[True]

print("Total passengers with age less than 30:")
print(passengers_less_than_30)

first_class_fare = df.groupby("pclass")["fare"].sum()[1]

print("\nTotal fare paid by first class passengers:")
print(first_class_fare)

survivors_by_class = (
    df[df["survived"] == "1"]
    .groupby("pclass")
    .size()
)

print("\nNumber of survivors in each passenger class:")
print(survivors_by_class)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from sklearn.datasets import load_wine

wine = load_wine()

df = pd.DataFrame(
    wine.data,
    columns=wine.feature_names
)

numeric_features = df.select_dtypes(include=np.number).columns
categorical_features = df.select_dtypes(exclude=np.number).columns

print("Number of Numeric Features:")
print(len(numeric_features))

print("\nNumber of Categorical Features:")
print(len(categorical_features))

corr_matrix = df.corr()

for i in corr_matrix.columns:
    for j in corr_matrix.columns:
        if i != j:
            if corr_matrix[i][j] > 0.9:
                print("\nHighly correlated attributes:", i, "and", j)

                df = df.drop(columns=[j])

                break

print("\nFive Number Summary:\n")

summary = df.describe().loc[
    ["min", "25%", "50%", "75%", "max"]
]

print(summary)

plt.figure(figsize=(14, 8))

sns.boxplot(data=df)

plt.xticks(rotation=90)

plt.title("Five Number Summary Visualization")

plt.show()
```
